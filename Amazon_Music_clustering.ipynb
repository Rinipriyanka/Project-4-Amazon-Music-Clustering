```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
music_df=pd.read_csv("/content/single_genre_artists.csv")
music_df.info()
```

```python
#Null values
music_df.isnull().sum()
```

```python
#Data Cleaning
#Dropping the columns

music_df=music_df.drop(columns=["id_songs","id_artists","name_artists"])
music_df.info()
```

```python
#FInding Duplicates

music_df.duplicated().sum()
music_df[music_df.duplicated()]
```

```python
#Dropping Duplicates

music_df=music_df.drop_duplicates()
music_df.duplicated().sum()
```

```python
music_df.info()
```

```python
music_df["name_song"]=(music_df["name_song"].str.lower().str.replace(r"[^a-zA-Z0-9\s]","",regex=True))
music_df["name_song"]
```

```python
music_df["genres"]=(music_df["genres"].str.lower().str.replace(r"[\[\]']","",regex=True).str.strip())
music_df["genres"]
```

```python
music_df = music_df.copy()
```

```python
music_df
```

```python

music_df["followers"]=music_df["followers"].astype("Int64")
music_df
```

```python
music_df.info()
```

```python
music_df["release_date"].head(10)
```

```python
music_df["release_date"].isna().sum()
```

```python
music_df["release_year"]=pd.to_datetime(music_df["release_date"],errors="coerce",format="mixed").dt.year
music_df
```

```python
music_df.info()
```

```python
#Feature Selection


features=["danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo",
    "duration_ms"]
X=music_df[features]
```

```python
#Standard Scaler
#puts different values into comparable scale

from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()

x_scaled=scaler.fit_transform(X)
```

```python
#K means

from sklearn.cluster import KMeans
inertia=[] #squared distance from the centroid

for k in range(2,11):
  kmeans=KMeans(n_clusters=k,random_state=42,n_init=10)
  kmeans.fit(x_scaled)
  inertia.append(kmeans.inertia_)
```

```python
#Plot the Elbow

import matplotlib.pyplot as plt
plt.plot(range(2,11),inertia,marker="o")
plt.xlabel("No of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
#Silhouette Score
#It measures how well-separated and cohesive the clusters are

from sklearn.metrics import silhouette_score
import numpy as np

sample_idx = np.random.RandomState(42).choice(
    len(x_scaled),
    size=10000,
    replace=False
)

for k in [4, 5]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(x_scaled)

    score = silhouette_score(
        x_scaled[sample_idx],
        labels[sample_idx]
    )

    print("K =", k, "Silhouette Score =", score)
```

```python
#Kmeans with cluster=4

kmeans=KMeans(n_clusters=4,random_state=42,n_init=10)
cluster_label=kmeans.fit_predict(x_scaled)
music_df["cluster"]=cluster_label

```

```python
music_df.cluster.value_counts()
```

```python
#Calculating average for each cluster

cluster_summary=music_df.groupby("cluster")[features].mean().round(2)
print(cluster_summary)
```

```python
#Davies Bouldin Index

from sklearn.metrics import davies_bouldin_score
db_score=davies_bouldin_score(x_scaled,cluster_label)
print("davies_bouldin_score",db_score)
```

```python
from sklearn.metrics import davies_bouldin_score

for k in [4, 5]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(x_scaled)

    score = davies_bouldin_score(x_scaled, labels)

    print("K =", k, "DBI =", score)
```

```python
#Visualization

#PCA
#used for visualizing 10 dimension data into 2 dimension

from sklearn.decomposition import PCA
pca=PCA()
x_pca=pca.fit_transform(x_scaled)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(x_pca[:,0],x_pca[:,1],c=cluster_label,s=5)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Scatter Plot")
plt.show()
```

```python
#Bar Chart

import pandas as pd
cluster_summary.plot(kind="bar",figsize=(10,8))
plt.xlabel("clusters")
plt.ylabel("Average")
plt.title("Avg values of clusters")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05,1))
plt.show()
```

```python
#standardize the cluster averages for visualization

cluster_summary_scaled=(cluster_summary-cluster_summary.mean())/cluster_summary.std()
cluster_summary_scaled.plot(kind="bar",figsize=(10,5))
plt.xlabel("Cluster")
plt.ylabel("Standardized Average")
plt.title("Average Feature Values by Cluster")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
#Heatmap

import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
sns.heatmap(cluster_summary_scaled,cmap="coolwarm",annot=True,fmt=".2f")
plt.title("cluster feature comparison")
plt.show()
```
